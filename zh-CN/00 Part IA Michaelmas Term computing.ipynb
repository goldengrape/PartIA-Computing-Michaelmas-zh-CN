## 课程结构

学习计算的最好方法是多做练习，并在需要时寻求帮助。本课程包括12个自学的交互式Jupyter笔记本'活动'，每个活动结束时都有练习。活动笔记本涵盖：

1. [变量，赋值和运算符优先级](01%20Variables,%20assignment%20and%20operator%20precedence.ipynb)
1. [控制语句](02%20Control%20statements.ipynb)
1. [类型，类型转换和浮点运算](03%20Types,%20type%20conversions%20and%20floating%20point%20arithmetic.ipynb)
1. [函数](04%20Functions.ipynb)
1. [库函数](05%20Library%20functions.ipynb)
1. [数据结构](06%20Data%20structures.ipynb)
1. [数值计算](07%20Numerical%20computation.ipynb)
1. [绘图](08%20Plotting.ipynb)
1. [错误处理和测试](09%20Error%20handling%20and%20testing.ipynb)
1. [算法](10%20Algorithms.ipynb)
1. [复杂度](11%20Complexity.ipynb)
1. [面向对象的设计](12%20Object-oriented%20design.ipynb)
1. [编译型和解释型语言](13%20Compiled%20and%20interpreted%20languages.ipynb) (可选)

这些笔记本旨在按顺序完成。要完成一项活动：

1. 通读笔记本，如果您想尝试一下，请更改部分程序以查看发生的情况 - 这将有助于您的理解。
1. 在进行下一个活动之前，请完成该笔记本的练习。

当您进行这些活动时，您可能需要复习一些早期的笔记本。

# 完成习题

每个笔记本的习题都在[Exercises](Exercises/) 目录中。

 - 使用Markdown单元格，并在适当的情况下使用LaTeX添加方程式，以描述您的程序正在执行的操作。
 - 对于较长的习题，您可能希望将程序分成多个单元格，并使用Markdown在代码单元格之前进行说明。
 - 大多数练习包括一些自动化测试，以提供有关你解答的正确与否的反馈。开始时，您不需要理解测试语法。
