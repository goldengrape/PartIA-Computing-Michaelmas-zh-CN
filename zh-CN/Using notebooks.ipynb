## LaTeX

Markdown单元支持排版数学公式的[LaTeX](https://www.latex-project.org/) 语法。 LaTeX是技术文档和数学展示的领先工具，它是免费的。

>学习LaTeX现在是Tripos后期的一项很好的投资。您可能需要尝试https://www.overleaf.com/ 以获取在线LaTeX环境。

要排版内联方程，请使用：
```
在这种情况下，所谓内联方程是 $\exp(-2x) \sin(3 x^{4})$ 。
```
这将显示为：

'在这种情况下，所谓内联方程是 $\exp(-2x) \sin(3 x^{4})$ 。'

对于显示的等式，来自

```
我们希望计算
当 $\beta = 4$ 时.
$$
f(x) = \beta x^{3} \int_{0}^{2} g(x) \, dx
$$
```
我们得到：

'我们希望计算
当 $\beta = 4$ 时.
$$
f(x) = \beta x^{3} \int_{0}^{2} g(x) \, dx
$$'

在线搜索以了解不同数学符号的LaTeX命令。如果您在笔记本中看到数学排版的示例，您还可以在Jupyter环境中双击它以查看语法。 https://notebooks.azure.com/library/1A-maths 上有很多例子。
